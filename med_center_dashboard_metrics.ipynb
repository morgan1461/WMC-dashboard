# Med Center Express Dashboard Metrics

### Overview

This module produces the dashboard metrics for the med center express route. 

The workflow is a Python translation of an existing R-based dashboard analysis previously used in the OSU Department of Transportation and Traffic Management (TTM), with improvements for modularity, readability, and scalability.

Note: This module contains code intended for internal TTM use in the future. For example, hourly breakdowns of bus headways, ignore this code while running the monthly metrics.

---

### Key Changes

This module is designed for med center express data **post December 2025**. This is due to the route changing and having 2 stops instead of 1. The John Herrick Transportation Hub stop was replaced with a stop at University Hospital on W 12th ave and a stop at Doan Hall on W 12th ave.

**NOTE:** For analysis purposes, the University Hosptial and Doan Hall stops were combined in both the capacity and travel time metrics. The headway metric is calculated the same, as it is based on the amount of time between busses at any given stop on the route, including Carmack 2, Carmack 3, University Hospital, and Doan Hall.

### Data Source

* Directory: `K:/AP/TTM/Data/WMC Dashbaord/BusState Cleaned`
* File format: `.csv`
* Naming convention:

  ```
  YYYY-MMM-busstate.csv
  ```

* `YYYY` = year 
* `MMM` = month abbreviation

Example: `2025-DEC-busstate.csv`

---

### Output

* Data is split into 3 files containing tables of metrics.

* Saved as `.csv` files:

* Example:

  ```
  1-Headway-AUG-2025.csv
  2-Capacity-AUG-2025.csv
  3-TravelTime-AUG-2025.csv
  ```

* Output directory:

  ```
  K:/AP/TTM/Data/WMC Dashboard/Dashboard Data/2025
  ```
